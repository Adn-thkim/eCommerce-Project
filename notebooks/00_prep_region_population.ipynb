```python
import pandas as pd
```

```python
pd.set_option("display.max_rows", 5000)
```

```python
region_df = pd.read_csv("../raw_data/행정구역_시군구_별_성별_인구수_2025.csv", encoding="euc-kr")

region_df = region_df.rename(columns={'행정구역(시군구)별' : "지역구", "항목" : "성별", "데이터" : "인구수"})
region_df["성별"] = region_df["성별"].apply(lambda x: x.replace("인구수 (명)", ""))
region_df = region_df.drop("시점", axis=1)

korea_cities = ["서울특별시", "부산광역시", "인천광역시", "광주광역시", "대전광역시", "울산광역시", "세종특별자치시", "경기도", "대구광역시",
                "강원특별자치도", "충청북도", "충청남도", "전북특별자치도", "전라남도", "경상북도", "경상남도", "제주특별자치도"]
region_df["시도"] = region_df["지역구"].apply(lambda x: x if x in korea_cities else "")
region_df["시도"] = region_df["시도"].replace("", pd.NA).ffill()

region_df = region_df[region_df["인구수"]>0]
region_df = region_df[region_df["지역구"]!=region_df["시도"]]

region_df.head()
```

```python
region_df.to_csv("../data/행정구역별_성별_인구통계_2025.csv", index=False)
```
